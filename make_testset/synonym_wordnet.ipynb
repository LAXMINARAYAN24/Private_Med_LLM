```python
import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet
import pandas as pd
pd.set_option('display.max_rows', None)
```

```python
syns=wordnet.synsets("fever.n.01")
for s in syns:
    print(s.name())
```

```python
des = pd.read_csv('/content/MedCAT_Descriptions.csv')
```

```python
des = des.drop_duplicates()
des = pd.DataFrame(des)
des.describe()
```

```python
des['synonym'] = ''
des['syn_len'] = 0
des['DESCRIPTION'] = des['DESCRIPTION'].apply(lambda x: x.lower())
des['synonym'] = des['DESCRIPTION'].apply(lambda x: list(set([i.lemmas()[0].name() for i in wordnet.synsets(x)])))
des['synonym'] = des['synonym'].apply(lambda x: [i.replace('_', ' ') for i in x])
des['syn_len'] = des['synonym'].apply(lambda x: len(x))
```

```python
des[des['syn_len']>0][:10]
```

```python
def check_first_item(synonyms, description, syn_len):
    # Ensure the synonym list is not empty, syn_len is 1, and the first item matches the DESCRIPTION
    if syn_len == 1 and synonyms and synonyms[0].lower() == description.lower():
        return True
    return False

```

```python
for index, row in des.iterrows():
    if check_first_item(row['synonym'], row['DESCRIPTION'], row['syn_len']):
        des.at[index, 'synonym'] = []

```

```python
non_empty_synonym_rows = des[des['synonym'].apply(lambda x: len(x) != 0)]
```

```python
non_empty_synonym_rows.to_csv('non_empty_synonym_rows.csv', index=False)
```
