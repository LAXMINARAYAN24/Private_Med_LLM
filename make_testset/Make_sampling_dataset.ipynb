```python
import os
from tqdm import tqdm
import pandas as pd
import pickle
tqdm.pandas()
```

```python


with open("data/df_train_prompt.pkl", "rb") as f:
    train_data = pickle.load(f)

with open("data/df_test_prompt.pkl", "rb") as f:
    test_data = pickle.load(f)

df = pd.read_csv('MIMIC/pesudo_mimic3_processed.csv')
icd_descript = pd.read_csv('MIMIC/ICD9_Descriptions.csv')

```

### Frequency

```python
train_data['condition'].values
```

```python
all_string = ''
for conditions in train_data['condition'].values:
    try:
        for i in conditions:
            all_string = all_string +', '+ i
    except:
        print(i)
```

```python
from collections import Counter
combined_condition = all_string.split(', ')
all_condition = Counter(combined_condition)

condition = list(all_condition.keys())
count = list(all_condition.values())

df_count = pd.DataFrame({'condition': condition, 'count': count})
```

```python
df_count = df_count.sort_values(by='count', ascending=False)
df_count
```

```python
df_count['condition'] = df_count['condition'].apply(lambda x: x.lower())
df_count=df_count.dropna()
df_count = df_count.drop_duplicates()
```

```python
total = df_count['count'].sum()
df_count['ratio'] = df_count['count']/total*100

df_count
```

```python
df_count.describe()
```

```python
print(df_count[:100]['ratio'].sum())
general_conditions = list(df_count[:100]['condition'])
```

```python
(general_conditions)
```

```python
df_train = train_data
df_train['overlap'] = df_train['condition'].apply(lambda x: sum(c.lower() in [word.lower() for word in x] for c in general_conditions))
df_train['overlap_ratio'] = df_train['overlap']/df_train['num']*100
df_train['a'] = 0
df_train
```

```python
df_train.describe()
```

```python
quantile_33 = df_train['overlap_ratio'].quantile(0.333)
quantile_67 = df_train['overlap_ratio'].quantile(0.667)
print(f"quantile_33: {quantile_33}, quantile_67: {quantile_67} ")

def classify_by_percentile(x):
    if x <= quantile_33:
        return 1
    elif x <= quantile_67:
        return 2
    else:
        return 3

df_train['a'] = df_train['overlap_ratio'].apply(classify_by_percentile)
```

```python
df_train = df_train[['SUBJECT_ID', 'name', 'a', 'condition','code','gender', 'num']]
df_train.columns = ['SUBJECT_ID', 'name', 'a', 'condition','code','gender', 'condition_nums']
```

```python
df_train_1 = df_train[df_train['a']==1].sample(n=1333, replace=False, random_state=42)[['SUBJECT_ID', 'name', 'a', 'condition','code','gender', 'condition_nums']]
df_train_2 = df_train[df_train['a']==2].sample(n=1333, replace=False, random_state=42)[['SUBJECT_ID', 'name', 'a', 'condition','code','gender', 'condition_nums']]
df_train_3 = df_train[df_train['a']==3].sample(n=1334, replace=False, random_state=42)[['SUBJECT_ID', 'name', 'a', 'condition','code','gender', 'condition_nums']]

sample_b = pd.concat([df_train_1,df_train_2,df_train_3], axis=0)
sample_b.a.value_counts()
```

```python
sample_b.to_csv('sample_frequency_4000.csv', index=False)
```
