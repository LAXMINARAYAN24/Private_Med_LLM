```python
def process_output(output):
    split_output = output.split('###Answer:', 1)
    if len(split_output) > 1:
        return split_output[1].strip()
    return output.strip()

def determine_answer(output):
    output_lower = output.lower()
    if 'female' in output_lower:
        return 'F'
    elif 'male' in output_lower:
        return 'M'
    else:
        if 'woman' in output_lower:
            return 'F'
        elif 'man' in output_lower:
            return 'M'
        print(output_lower)
        return 'Other'
```

```python
model = 'meditron'
sample = 'frequency'
```

```python
rand = pd.read_csv(f'test_result/gender/{model}_sample_{sample}_gender_fake_False.csv')
rand['output'] = rand['output'].apply(process_output)
rand['answer'] = rand['output'].apply(determine_answer)

correct_predictions = (rand['gender'] == rand['answer']).sum()
total_predictions = len(rand)
accuracy = correct_predictions / total_predictions
```
