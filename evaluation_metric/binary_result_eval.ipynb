```python
import numpy as np
import pandas as pd
import pickle
import random
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def process_output(output):
    split_output = output.split('###Answer:', 1)
    if len(split_output) > 1:
        return split_output[1].strip()
    return output.strip()

def determine_answer(output):
    output_lower = output

    if 'Yes' in output_lower[:10]:
        return 'Y'
    elif 'No' in output_lower[:10]:
        return 'N'
    else:
        if'not' in output_lower:
            return 'N'
        else:
            return 'Y' 
```

```python
model = 'meditron'
data = 'random'
```

```python
rand = pd.read_csv(f'test_result/binary/{model}_sample_random_binary_fake_False.csv')
frequency = pd.read_csv(f'test_result/binary/{model}_sample_frequency_binary_fake_False.csv')
length = pd.read_csv(f'test_result/binary/{model}_sample_length_binary_fake_False.csv')
```

```python
def eval_rand(rand, sample, fake=False):
    rand['output'] = rand['output'].apply(process_output)
    rand['answer'] = rand['output'].apply(determine_answer)

    n_answers = rand[rand['answer'] == 'N']
    y_answers = rand[rand['answer'] == 'Y']
    n_answers_with_not = rand[(rand['output'].str.contains('not'))&(rand['answer']=='Y')]


    plt.figure(figsize=(8, 8))
    plt.pie([len(n_answers),len(y_answers)], labels=['N', 'Y'], autopct='%1.1f%%', startangle=90)
    plt.title(f'{sample} of N and Y Answers (fake={fake})')
    plt.axis('equal')
    plt.show()

```

```python
eval_rand(rand, 'random')
eval_rand(frequency, 'frequency')
eval_rand(length, 'length')
```

```python
rand_fake = pd.read_csv(f'test_result/binary/{model}_sample_random_binary_fake_True.csv')
frequency_fake = pd.read_csv(f'test_result/binary/{model}_sample_frequency_binary_fake_True.csv')
length_fake = pd.read_csv(f'test_result/binary/{model}_sample_length_binary_fake_True.csv')
```

```python
eval_rand(rand_fake, 'random', True)
eval_rand(frequency_fake, 'frequency', True)
eval_rand(length_fake, 'length', True)
```
