```python
import numpy as np
import pandas as pd
import pickle
import random
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def process_output(output):
    split_output = output.split('###Answer:', 1)
    if len(split_output) > 1:
        return split_output[1].strip()
    return output.strip()

def determine_answer(output):
    return output[0]


def eval_accuracy(rand, sample):
    correct_predictions = (rand['correct_option'] == rand['answer']).sum()
    total_predictions = len(rand)
    accuracy = correct_predictions / total_predictions

    return accuracy
```

```python

rand['output'] = rand['output'].apply(process_output)
rand['answer'] = rand['output'].apply(determine_answer)

length['output'] = length['output'].apply(process_output)
length['answer'] = length['output'].apply(determine_answer)

freq['output'] = freq['output'].apply(process_output)
freq['answer'] = freq['output'].apply(determine_answer)
```

```python
rand_accuracy = eval_accuracy(rand, 'random')
leng_accuracy = eval_accuracy(length, 'length')
freq_accuracy = eval_accuracy(freq, 'frequency')
```
